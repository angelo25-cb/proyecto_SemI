```python
import pandas as pd

df = pd.read_excel("../data/dataset_sintetico_demanda_lima.xlsx")

df["fecha"] = pd.to_datetime(df["fecha"])

df["dia_semana_num"] = df["fecha"].dt.dayofweek

df["es_fin_semana"] = (
    df["dia_semana_num"].isin([5,6]).astype(int)
)

df.head()
```

```python
df = pd.get_dummies(
    df,
    columns=["dia_semana"]
)
```

```python
df["demanda_lag_1"] = df["demanda_real"].shift(1)
df["demanda_lag_7"] = df["demanda_real"].shift(7)

df.dropna(inplace=True)
```
